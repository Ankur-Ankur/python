## Topic 5: Loops

1. Calculate total revenue from a list of order dictionaries.
2. Count paid orders and collect the IDs of unpaid orders.
3. Validate records in a loop and collect IDs whose amount is missing or non-positive.

```python
# Answers 1-3
# Write your answers here
#Calculate total revenue from a list of order dictionaries
orders = [
    {"order_id": 1, "revenue": 1000},
    {"order_id": 2, "revenue": 2750},
    {"order_id": 3, "revenue": 1750}
]

total_revenue = 0

for order in orders:
    total_revenue = total_revenue + order["revenue"]

print(total_revenue)
```

```python
#Count paid orders and collect the IDs of unpaid orders.

orders = [
    {"order_id": 1, "revenue": 1000, "status": "paid"},
    {"order_id": 2, "revenue": 2750, "status": "unpaid"},
    {"order_id": 3, "revenue": 1750, "status": "paid"}
]
paid_orders = 0
unpaid_orders = []
for order in orders:
    if order ["status"] == "paid":
        paid_orders += 1
    else:
        unpaid_orders.append(order["order_id"])

print("Paid orders:", paid_orders)
print("Unpaid ids:", unpaid_orders)                           
```

```python
#Validate records in a loop and collect IDs whose amount is missing or non-positive.

orders = [
    {"order_id": 1, "revenue": 1000},
    {"order_id": 2, "revenue": 2750},
    {"order_id": 3, "revenue": 1750}
]
invalid_orders = []
for order in orders:
   amount = order.get("revenue")
   if amount is None or amount <= 0:
     invalid_orders.append(order["order_id"])

print(invalid_orders)    

```

## Topic 6: Functions

1. Write `net_price(price, discount_rate=0)` with validation for a rate from 0 to 1.
2. Write `active_employee_names(records)` that returns names for active employees.
3. Write `summarize_values(values)` returning count, total, and average.

```python
# Answers 1-3
# Write your answers here

def net_price(price, discount_rate=0):
    if discount_rate < 0 or discount_rate > 1:
        return "Invalid discount rate"

    final_price = price * (1 - discount_rate)
    return final_price
print(net_price(1500,0.20))
```

```python
#Write active_employee_names(records) that returns names for active employees.

employees = {
    'emp_1': {'name': 'John', 'salary': 50000, 'bonus': 5000, 'status': 'Active'},
    'emp_2': {'name': 'Alice', 'salary': 45000, 'bonus': 3000, 'status': 'Inactive'},
    'emp_3': {'name': 'Sam', 'salary': 52000, 'bonus': 4000, 'status': 'Active'},
}



def active_employee_names(records):
    active_names = []
    for employee in records.values():
        if employee["status"] == "Active":
            active_names.append(employee["name"])
    return active_names
print(active_employee_names(employees))
```

```python
#Write summarize_values(values) returning count, total, and average.
values = [10, 20, 30, 40]

def summarize_values(values):
    count = len(values)
    total = sum(values)
    average = total / count

    return count, total, average

print(summarize_values(values))
```
