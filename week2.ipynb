## Topic 5: Loops

1. Calculate the total of `[25, 40, 15, 20]`.
2. Count values at least 20 in the same list.
3. Print each item in `['paid', 'pending', 'paid']` with its position using `enumerate()`.

```python
# Answers 1-3
# Write your answers here

x = [25,40,15,20]
print(sum(x))
```

```python
x = [25,40,15,20]

count = 0
for number in x:
    if number >=20:
        count = count+1
print(count)        
```

```python
status = ["paid", "pending", "paid"]

for position, item in enumerate(status):
    print(position, item)
```

## Topic 6: Functions

1. Write `square(number)` that returns the number squared.
2. Write `full_name(first, last)` that returns a combined string.
3. Write `shipping_cost(amount, free_limit=100)` that returns 0 when the amount reaches the limit, otherwise 10.

```python
# Answers 1-3
# Write your answers here
def square(number):
    return number * number

print(square(5))

```

```python
def full_name(first, last):
    return first + " " + last

print(full_name("Ankur", "Kamboj"))
```

```python
def shipping_cost(amount, free_limit=100):
    if amount >= free_limit:
        return 0
    else:
        return 10

print(shipping_cost(150))
```
